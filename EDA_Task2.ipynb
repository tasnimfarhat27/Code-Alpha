```python
import pandas as pd
import matplotlib.pyplot as plt

# Load Dataset
df = pd.read_csv("scraped_job_data.csv")

print("=" * 50)
print("CODEALPHA TASK 2 - EXPLORATORY DATA ANALYSIS")
print("=" * 50)
```

```python
# -----------------------------------
# 1. Basic Information
# -----------------------------------

print("\nDataset Shape:")
print(df.shape)

print("\nDataset Information:")
print(df.info())

print("\nFirst 5 Rows:")
print(df.head())
```

```python
# -----------------------------------
# 2. Missing Values Analysis
# -----------------------------------

print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# -----------------------------------
# 3. Duplicate Records
# -----------------------------------

duplicates = df.duplicated().sum()

print(f"\nDuplicate Records: {duplicates}")
```

```python
# -----------------------------------
# 5. Top Hiring Companies
# -----------------------------------

top_companies = df["Company"].value_counts().head(10)

print("\nTop Hiring Companies:")
print(top_companies)
```

```python
# -----------------------------------
# 6. Top Job Locations
# -----------------------------------

top_locations = df["Location"].value_counts().head(10)

print("\nTop Locations:")
print(top_locations)
```

```python
# -----------------------------------
# 8. Date Analysis
# -----------------------------------

df["Date Posted"] = pd.to_datetime(
    df["Date Posted"],
    errors="coerce"
)

print("\nDate Range:")
print("Earliest Date:", df["Date Posted"].min())
print("Latest Date:", df["Date Posted"].max())
```

```python
# -----------------------------------
# 9. Summary Statistics
# -----------------------------------

print("\nSummary Statistics:")
print(df.describe(include="all"))

```

```python
# -----------------------------------
# 10. Visualization - Top Companies
# -----------------------------------

plt.figure(figsize=(10,5))
top_companies.plot(kind="bar")
plt.title("Top Hiring Companies")
plt.xlabel("Company")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# -----------------------------------
# 11. Visualization - Top Locations
# -----------------------------------

plt.figure(figsize=(10,5))
top_locations.plot(kind="bar")
plt.title("Top Job Locations")
plt.xlabel("Location")
plt.ylabel("Number of Jobs")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
