```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

# 1. Target URL (a sandbox site designed for scraping practice)
url = "https://books.toscrape.com/"

# 2. Fetch HTML content
response = requests.get(url)

if response.status_code == 200:
    # 3. Parse HTML page
    soup = BeautifulSoup(response.text, "html.parser")
    books = soup.find_all("article", class_="product_pod")
    
    scraped_data = []

    # 4. Extract data fields
    for book in books:
        title = book.h3.a["title"]
        price = book.find("p", class_="price_color").text
        availability = book.find("p", class_="instock availability").text.strip()
        
        scraped_data.append({
            "Title": title,
            "Price": price,
            "Availability": availability
        })

    # 5. Convert to Pandas DataFrame
    df = pd.DataFrame(scraped_data)
    
    # Save locally as CSV
    df.to_csv("scraped_books.csv", index=False)
    print("Scraping complete! Displaying first 5 rows below:")
    
    # Display table directly in Jupyter
    display(df.head())
else:
    print(f"Failed to fetch site. Status code: {response.status_code}")
```
